## Data Understanding

This notebook is used to load and inspect the official Kaggle competition datasets before data cleaning, preprocessing, and model development.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

print("Required libraries imported successfully")
```

```python
current_folder = Path.cwd()

print("Current working folder")
print(current_folder)
```

```python
if (current_folder / "data").exists():
    project_folder = current_folder
else:
    project_folder = current_folder.parent

print("Project folder")
print(project_folder)
```

```python
data_folder = project_folder / "data" / "kaggle" / "raw"

print("Dataset folder")
print(data_folder)

print("Folder exists")
print(data_folder.exists())
```

```python
dataset_files = list(data_folder.iterdir())

print("Files inside the dataset folder")

for file in dataset_files:
    print(file.name)
```

```python
train_file = data_folder / "train.csv"
test_file = data_folder / "test.csv"
submission_file = data_folder / "sample_submission.csv"

print("Training file exists", train_file.exists())
print("Testing file exists", test_file.exists())
print("Submission file exists", submission_file.exists())
```

```python
train_data = pd.read_csv(train_file)
test_data = pd.read_csv(test_file)
sample_submission = pd.read_csv(submission_file)

print("All datasets loaded successfully")
```

```python
print("Training dataset shape", train_data.shape)
print("Testing dataset shape", test_data.shape)
print("Sample submission shape", sample_submission.shape)
```

```python
target_column = "health_condition"

print("Target column", target_column)
print("Target exists in training data", target_column in train_data.columns)
print("Target exists in testing data", target_column in test_data.columns)
```

```python
train_missing_values = train_data.isnull().sum()

print("Missing values in training data")
print(train_missing_values)
```

```python
test_missing_values = test_data.isnull().sum()

print("Missing values in testing data")
print(test_missing_values)
```

```python
train_missing_percentage = train_data.isnull().mean() * 100
test_missing_percentage = test_data.isnull().mean() * 100

print("Training missing-value percentages")
print(train_missing_percentage.round(2))

print()

print("Testing missing-value percentages")
print(test_missing_percentage.round(2))
```

```python
training_duplicates = train_data.duplicated().sum()
testing_duplicates = test_data.duplicated().sum()

print("Duplicated training rows", training_duplicates)
print("Duplicated testing rows", testing_duplicates)
```

```python
target_counts = train_data[target_column].value_counts()

print("Target class counts")
print(target_counts)
```

```python
target_percentages = train_data[target_column].value_counts(normalize=True) * 100

print("Target class percentages")
print(target_percentages.round(2))
```

```python
numerical_columns = train_data.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns = train_data.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical columns")
print(numerical_columns)

print()

print("Categorical columns")
print(categorical_columns)
```

```python
categorical_feature_columns = categorical_columns.copy()

if target_column in categorical_feature_columns:
    categorical_feature_columns.remove(target_column)

print("Categorical feature columns")
print(categorical_feature_columns)
```

```python
training_features = set(train_data.columns) - {target_column}
testing_features = set(test_data.columns)

missing_from_test = training_features - testing_features
extra_in_test = testing_features - training_features

print("Training features missing from testing data")
print(missing_from_test)

print()

print("Extra features in testing data")
print(extra_in_test)
```

## Initial observations

The training dataset contains 690,088 rows and 15 columns.

The testing dataset contains 295,753 rows and 14 columns.

The target variable is `health_condition`.

This is a supervised multiclass classification problem with the classes `at-risk`, `unhealthy`, and `fit`.

The raw training dataset contains 8 numerical columns including `id` and 6 categorical input features.

The training dataset contains 449,496 missing values and the testing dataset contains 192,642 missing values.

No duplicated rows were found in either dataset.

The target is highly imbalanced because `at-risk` represents 85.87 percent, `unhealthy` represents 8.36 percent, and `fit` represents 5.77 percent.
